# Series 2.8 — Multi-Agent Orchestration

**Why AI Fails? — Engineering Lab**

---

> Enterprise AI systems scale through **coordination** rather than intelligence.

**Scenario:** 500 enterprise software build requests — four orchestration strategies, ten specialized agents, shared memory bus.

**Core lesson:** One agent can answer. A coordinated team of specialized agents can **build**.


## End-to-End Scenario

**What this lab does from start to finish**

A product manager submits an enterprise software request: *"Design and build a secure REST API for inventory management with authentication, database schema, tests, and deployment docs."* One general-purpose agent would produce a shallow answer. This lab coordinates **ten specialized agents** through a planner, scheduler, shared memory bus, and optional reviewer — across **500 benchmark requests**.

### Step-by-step flow

1. **Receive an enterprise request** — `tasks.py` provides 500 requests across 8 categories (greenfield apps, API design, security audits, migrations, etc.).
2. **Plan the work** — **Planner Agent** (`planner.py`) decomposes the request into domain tasks with dependencies (architecture → backend → database → security → tests → docs).
3. **Schedule execution** — `scheduler.py` runs tasks **sequentially** (one after another) or in **parallel waves** (independent tasks simultaneously).
4. **Execute with specialized agents** — `agents.py` pool:
   - Architecture, Backend, Frontend, Database, Security, Testing, DevOps, Documentation (+ Planner, Reviewer)
   - Each agent reads context from and writes outputs to **SharedMemory** (`shared_memory.py`) — agents never talk directly to each other.
5. **Aggregate results** — `orchestrator.py` merges agent outputs into a unified deliverable.
6. **Optional review pass** — **Reviewer Agent** (`reviewer.py`) checks consistency, finds gaps, and triggers **one rework iteration** if needed.
7. **Run four orchestration strategies:**
   - **Single** — one general agent handles everything
   - **Sequential** — planner → agents in dependency order
   - **Parallel** — independent agents in parallel waves
   - **Parallel + Reviewer** — parallel execution + validation pass
8. **Evaluate output** — `evaluator.py` scores quality, consistency, and task completion.
9. **Benchmark all strategies** — `benchmark.py` compares latency, quality score, and completion rate (dry-run uses tuned simulation targets).
10. **Live demo cell** — run `python series-2.8/app.py --dry-run --requests 50`.

### What you should observe

| Strategy | Latency (dry-run) | Quality (dry-run) | Coverage |
|----------|-------------------|-------------------|----------|
| Single | ~18s | 0.84 | One perspective |
| Sequential | ~14s | 0.91 | Full domain depth |
| Parallel | ~8s | 0.92 | Fast + broad |
| Parallel + Reviewer | ~10s | 0.97 | Highest quality |

**Takeaway:** One intelligent agent can **answer**. A coordinated team of specialized agents — with a planner, scheduler, shared memory, and reviewer — can **build**.


## 1. The Problem

| Single general agent | Multi-agent orchestration |
|----------------------|---------------------------|
| One prompt, one response | Planner decomposes into domain tasks |
| Shallow coverage of architecture, security, tests | Each domain gets a specialist |
| No dependency management | Scheduler runs sequential or parallel waves |
| Agents talk through prompt bloat | Agents communicate via **Shared Memory** only |
| No quality gate | Reviewer validates + one rework iteration |

**Expected dry-run results:**

```
SINGLE AGENT         →  ~18s latency, quality ~0.84, 50% task completion
SEQUENTIAL           →  ~14s latency, quality ~0.91, full completion
PARALLEL             →  ~8s latency, quality ~0.92, full completion
PARALLEL + REVIEWER  →  ~10s latency, quality ~0.97, enterprise pick
```


## 2. What is Multi-Agent Orchestration?

**Multi-agent orchestration** coordinates **specialized AI agents** through a central orchestration layer to solve complex enterprise tasks.

### Definition

```
Orchestration = request → planner → scheduler → agents → shared memory
                → aggregator → (reviewer) → final response
```

### Key principle

Agents **never communicate directly**. They read and write **Shared Memory** only.

### Agent pool

| Agent | Domain |
|-------|--------|
| Planner | Task decomposition, dependencies |
| Architecture | System design, API boundaries |
| Backend | REST APIs, business logic |
| Frontend | UI components |
| Database | Schema, indexes, SQL |
| Security | Auth, authorization, vulnerabilities |
| Testing | Unit + integration tests |
| DevOps | Docker, Kubernetes, CI/CD |
| Documentation | README, API docs |
| Reviewer | Consistency, gaps, validation |

### Four strategies

| Strategy | CLI | Behavior |
|----------|-----|----------|
| Single | `--strategy single` | One general agent (baseline) |
| Sequential | `--strategy sequential` | Agents run in dependency order |
| Parallel | `--strategy parallel` | Independent agents in parallel waves |
| Reviewer | `--strategy reviewer` | Parallel + review + one rework pass |

## 5. The Orchestration Pipeline (`orchestrator.py`)

```
User Request
    ↓
Planner Agent → Task Decomposition
    ↓
Scheduler (sequential or parallel waves)
    ↓
Specialized Agents (read/write Shared Memory)
    ↓
Result Aggregator
    ↓
Reviewer Agent (reviewer strategy only)
    ↓
Final Response
```

**Example task graph** for "Build a production-ready banking application":

```
Architecture → Database → Backend → Frontend
                    ↘         ↓
                  Security  Testing → DevOps → Documentation
```


## 6. Three Layers of Orchestration Engineering

### Layer 1 — Plan and schedule (decompose before executing)

The **Planner** breaks one enterprise request into dependent tasks. The **Scheduler** orders them sequentially or groups parallel waves.

---

### Layer 2 — Coordinate through Shared Memory

Every agent reads prior outputs from shared memory and writes its deliverable. No direct agent-to-agent messages — prevents inconsistent handoffs.

Memory keys: `architecture`, `database`, `backend`, `frontend`, `security`, `testing`, `devops`, `documentation`, `review`

---

### Layer 3 — Measure quality per unit time

| Metric | What it tells you |
|--------|-------------------|
| **Task Completion** | Required domains delivered |
| **Consistency Score** | Cross-agent alignment |
| **Security Score** | Regulated-request coverage |
| **Review Score** | Reviewer pass quality |
| **Overall Quality** | Weighted composite |
| **Latency** | Sequential vs parallel tradeoff |
